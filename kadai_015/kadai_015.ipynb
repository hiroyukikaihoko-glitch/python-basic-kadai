```python
# 1. Humanクラスを定義
class Human:
    # コンストラクタ（初期化メソッド）でnameとageを定義
    def __init__(self, name, age):
        self.name = name  # 名前を表す属性
        self.age = age  # 年齢を表す属性

    # nameとageを標準出力するメソッド
    def printinfo(self):
        print(f"名前: {self.name}")
        print(f"年齢: {self.age}歳")


# 2. Humanクラスのインスタンスを作成し、変数に代入
person = Human(name="侍 太郎", age=25)

# 3. メソッドを実行して属性を出力
person.printinfo()
```
