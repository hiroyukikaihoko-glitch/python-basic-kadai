```python
array = ["水","金","地","火","木","土","天","海","冥"]
for array_name in array:
    #対応する要素を出力
    print(array_name)

array = ["水","金","地","火","木","土","天","海","冥"]
i = 0  # 0番目のインデックスから
while i < 9:   # iが配列の要素数（9）より小さい間ループ
    print(array[i])         # 配列の要素を出力
    i += 1                  # iの値を1つ増やす
```
