```python
#ランダムな整数を利用するために、randomモジュールをインポート
import random

# 変数varに0～15までのランダムな整数を代入する
var = random.randint(0, 15)
#print(var) #事前確認

# 条件分岐（3と5の両方の倍数 = 15の倍数 から先に判定）
if var % 3 == 0 and var % 5 == 0:
    print("Fizz")
elif var % 5 == 0:
    print("Buzz")
elif var % 3 == 0:
    print("FizzBuzz")
else:
    print(var)
```
